```python
from googletrans import Translator
import pandas as pd
df = pd.read_excel(r'C:\Users\va185069\OneDrive - NCR ATLEOS\Desktop\Fail_analysis\Part_Details.xlsx')
df = df.head(10)

def trans(txt):
    translator = Translator()
    kd = translator.translate(txt)
    return kd.text

df['trans'] = df['Problem Summary Text'].apply(trans)

df.to_excel(r'C:\Users\va185069\OneDrive - NCR ATLEOS\Desktop\Fail_analysis\'Part_Details_12.xlsx')
```
